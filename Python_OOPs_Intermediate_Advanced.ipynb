### 4 . Student Score System

Create a class Student:

Private variable __score

Method set_score(score)

Method get_score()

Score must be between 0 and 100

```python
class Student:
  def __init__(self, score):
    if 0 <= score <= 100:
        self.__score = score
    else:
        print("Score must be between 0 and 100")
        self.__score = 0  
  def set_score(self, score):
    if 0 <= score <= 100:
      self.__score = score
    else:
      print("Score must be between 0 and 100")
  def get_score(self):
      return self.__score
      
student_1 = Student(123)
print(student_1.get_score())
student_1.set_score(85)
print(student_1.get_score())
```

### 5 . Simple Wallet

Create a class Wallet:

Private variable __balance

Method deposit(amount)

Method get_balance()

Deposit amount must be positive

```python
class Wallet:
   def __init__(self,balance):
       if balance>=0:
        self.__balance = balance
       else :
        print("Amount should be positive")
        self.__balance=0
   def deposit(self,amount):
       if amount>0:
        self.__balance +=amount
       else :
        print("Amount should be positive")
   def get_balance(self):
     return self.__balance
```

```python
wallet_1 = Wallet(-1)
print(wallet_1.get_balance())

wallet_1.deposit(500)
print(wallet_1.get_balance())
```

## II . Abstraction

```python
from abc import ABC ,abstractmethod

class Car(ABC):
    @abstractmethod
    def moveforward(self):
        pass
    @abstractmethod
    def backward(self):
        pass
    @abstractmethod
    def fm (self):
        pass
class innova(Car):
    def moveforward(self):
      print ("Innova is moving forward !")
    def backward(self):
      print ("Innova is moving backward !")
    def fm(self):
      print ("Innova is playing fm !")
class swift(Car):
    def moveforward(self):
      print ("swift is moving forward !")
    def backward(self):
      print ("swift is moving backward !")
    def fm(self):
      print ("swift is playing fm !")    
innova_1=innova()
innova_1.backward()
swift_1=swift()
swift_1.moveforward()
```

### 1 . Shape Area

Create an abstract class Shape:

Abstract method area()

Create a child class Square:

Constructor takes side

Implement area() to return area

```python
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Square(Shape):
    def __init__(self,side):
     self.side=side
    def area(self):
        return self.side*self.side
     
```

```python
sq = Square(4)
print(sq.area())
```

### 3 . Multilevel Inheritance

### 4 . Hierarchical Inheritance
#### Create Shape as a parent class and Circle and Rectangle as child classes using hierarchical inheritance and call their methods.

```python
class Shape():
    def draw(self):
        print("Drawing Shape")
class Circle(Shape):
    def circle_area(self):
        print("Circle area calculated")
class Rectangle(Shape):
    def rectangle_area(self):
        print("Rectangle area calculated")
circle=Circle()
rectangle=Rectangle()
circle.draw()
circle.circle_area()
rectangle.draw()
rectangle.rectangle_area()
```

### 1 . Create two classes Dog and Cat with the same method sound() that prints “Dog Barks” and “Cat Meows”, then create objects and call the method.

### 2 . Create three classes Car, Bike, and Bus with the same method move() that prints “Car is driving”, “Bike is riding”, and “Bus is transporting passengers”, store the objects in a list and call the method using a loop.

```python
class Car():
    def move(self):
        print("Car is driving")
class Bike():
    def move(self):
        print("Bike is riding")
class Bus():
    def move(self):
        print("Bus is transporting passengers")
vehicles=Car(),Bike(),Bus()
for vehicle in vehicles:
    vehicle.move()
```

### 3 . Create a parent class Shape with method area(), then create child classes Square and Rectangle that override area() to calculate side × side and length × breadth, then create objects and call area().

```python
class Shape():
    def area(self):
        pass
        
class Square(Shape):
    def __init__(self,side):
        self.side=side
        
    def area(self):
        print(self.side*self.side)
        
class Rectangle(Shape):
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth   
        
    def area(self):
        print(self.length*self.breadth)
        
square=Square(5)
rectangle=Rectangle(20,10)

square.area()
rectangle.area()
```
