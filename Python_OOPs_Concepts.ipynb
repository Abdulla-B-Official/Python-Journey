# 4 Pillars of OOP — One by One, Like a Story
**Imagine this scenario:**

You go to an ATM:

You enter PIN → bank checks it secretly

You press Withdraw → money comes out

You press Check Balance → it shows your money

A Premium user uses same ATM but gets cashback

ATM has many internal things but you don’t see them

This is exactly what OOP means in programming — writing code like real life.


## 1️⃣ Encapsulation — “I hide important data”
**Simple meaning:**

We hide sensitive details so no one can change or steal them directly.

**In code:**
self.__pin = pin


The __ means: This data is private, don’t touch it from outside

**Beginner takeaway:**

Encapsulation = data protection


## 2️⃣ Abstraction — “You don’t need to know the details, just use it”
**Simple meaning:**

You use something without understanding the complex internal parts.

**Example:**

You press withdraw, but you don’t know:

How ATM talks to bank

How server checks balance

How cash machine works inside

**In code:**
@abstractmethod
def withdraw():
    pass


**This means:**

“Every user can withdraw, but each type must explain how it works themselves”

**Beginner takeaway:**

Abstraction = hide complexity, show only actions

## 3️⃣ Inheritance — “I can reuse code from another class”
**Simple meaning:**

A child can get features from a parent.

**Example:**

Parent: BankUser (has PIN check, balance check)

Child 1: NormalUser (uses parent powers + withdraws normally)

Child 2: PremiumUser (uses parent powers + gets cashback too)

**In code:**
class PremiumUser(BankUser):


**This means:**

PremiumUser gets everything BankUser has, no need to rewrite it

**Beginner takeaway:**

Inheritance = reuse, don’t repeat

## 4️⃣ Polymorphism — “Same action name, different result”
**Simple meaning:**

Same function name can do different things for different users.

**Example:**

Both users call:

withdraw()

But:

NormalUser → money deducted normally

PremiumUser → money deducted + bonus cashback

**Beginner takeaway:**

Polymorphism = one name, many behaviors


```python
from abc import ABC, abstractmethod

class BankUser(ABC):
    def __init__(self, name, pin, balance):
        self.name = name
        self.__pin = pin  # encapsulation
        self.__balance = balance  # encapsulation

    def verify_pin(self, attempt):
        return attempt == self.__pin  # abstraction of pin checking

    @abstractmethod
    def withdraw(self, attempt, amount):
        pass  # abstraction: child must define this

    def check_balance(self, attempt):
        if self.verify_pin(attempt):
            return f"{self.name}, your balance is: {self.__balance}"
        return "Wrong PIN!"

    def _deduct(self, amount):
        self.__balance -= amount

    def _balance(self):
        return self.__balance


class NormalUser(BankUser):  # inheritance
    def withdraw(self, attempt, amount):
        if self.verify_pin(attempt):
            if amount <= self._balance():
                self._deduct(amount)
                return f"{self.name} withdrew {amount}. Left: {self._balance()}"
            return "Insufficient funds"
        return "Wrong PIN"


class PremiumUser(BankUser):  # inheritance
    def withdraw(self, attempt, amount):
        if self.verify_pin(attempt):
            cashback = 500  # premium benefit
            if amount <= self._balance():
                self._deduct(amount - cashback)
                return f"{self.name} premium withdrew! Cashback {cashback}. Left: {self._balance()}"
            return "Insufficient funds"
        return "Wrong PIN"


# Objects
u1 = NormalUser("Fariq", 1234, 20000)
u2 = PremiumUser("Fariq Premium", 4321, 50000)

# Polymorphism: same method, different output
print(u1.withdraw(1234, 5000))
print(u2.withdraw(4321, 10000))
print(u1.check_balance(1234))
```
