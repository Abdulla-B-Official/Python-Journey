## 8. BREAK

### Example 1

#### Write a program using a while loop with break:
1. Keeps asking the user to enter a number.
2. If the user enters 0,stop the loop using break.
3. Otherwise,keep adding the entered numbers.
4. Finally,print the total sum.

```python
total=0
while True:
 user_input=int(input("enter a number :"))
 if user_input==0:
     break
 total=total+user_input
print("Total sum is",total)
```

### Example 2

#### Write a program using while loop with break:
1. Keeps asking the user to enter a number.
2. If the user enters a negative number,stop the loop using break.
3. count how many positive numbers were entered.
4. Print the count at the end.

```python
count=0
while True:
 user_input=int(input("Enter the number :"))
 if user_input < 0:
      break
 count+=1
print("Count of Positive numbers :",count)
```

## 9. CONTINUE

### Example 1

#### Write a program with For loop with Break and Continue 
1. Loops through numbers from  1 to 10.
2. If the number is 5 , skip it using CONTINUE.
3. If the number is 8 , stop it using BREAK.
4. Print all other numbers

```python

for i in range(1,10+1):
    if i==5:
     continue
    if i==8:
     break 
    print(i)
```
