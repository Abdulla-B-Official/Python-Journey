### Positional Arguments

```python
def get_details(name,place,expected_price):
    print("Client Name:",name)
    print("Place:",place)
    print("Expected_price:",expected_price)
```

```python
get_details("fariq","OOTY",95)
```

```python
get_details(place="chennai",expected_price="32",name="Abd ")
```

### Default Parameters & Arguments

```python
def get_details_chennai(name,expected_price,place="Chennai"):
    print("Client Name:",name)
    print("Place:",place)
    print("Expected_price:",expected_price)
```

```python
get_details_chennai("Fari",28)
```

```python
get_details_chennai("Fari",28,place="Banglore")
```

#### In one function , We can perform only any of the one in these two type (default or length)

### Variable length Argument

```python
def get_details(name,place,*expected_price):
    print("Client Name:",name)
    print("Place:",place)
    print("Expected_price:",expected_price)
```

```python
get_details("fari","ooty",20,30,40,50)
```

### Keyword Argument

```python
def get_details(name,place,**expected_price):
    print("Client Name:",name)
    print("Place:",place)
    print("Expected_price:",expected_price)
```

```python
get_details("fari","ooty",ask_1=20,ask_2=30,ask_3=40,ask_4=50)
```

## 5.4 Challenge-2 to find the factors of a number

```python
user_input=int(input("Enter the number"))

for i in range(1,user_input+1):
    if user_input%i==0:
       print(i)
```

```python
user_input=int(input("Enter the number"))
result=[]
for i in range(1,user_input+1):
    if user_input%i==0:
        result.append(i)
print(f"The factors of {user_input} is {result} ")
```
