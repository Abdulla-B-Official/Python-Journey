# Intermidiate Data Visualization

## Import Necesary Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
mtcars_data=pd.read_csv("mt_cars.csv")
mtcars_data.head()
```

### Must-NO
Should have the clarity of **Discrete and Continues**

## 1. Univariate Analysis

**Important charts:**

* Histogram
* Countplot
* Pie Chart

### 1.1 Histogram 
If you wanted to understand the distribution of one **Continues** feature you'l be going for **HISTOGRAM**

```python
plt.hist(mtcars_data.mpg)
plt.title(" MPG DISTRIBUTION ")
plt.xlabel("MPG")
plt.ylabel("Frequency")
plt.show()
```

### 1.2 Countplot - 
If you wanted to understand the distribution of one **Discrete** feature you'l be going for **COUNTPLOT**

```python
mtcars_data.head()
```

#### Note :
1. Matplotlib doesn't have charts which Seaborn has.
2. Seaborn will not have some charts which Matplotlib has.
3. Matplotlib and Seaborn will have same charts.

```python
sns.countplot(data=mtcars_data, x="am",y=None,)
plt.title("Automatic vs Manual Car Distribution")
plt.show()
```

### 1.3 Pie Chart
If you wanted to understand the distribution of one **Discrete** Along with its value and count with percentage wise you'l be going for **Pie Chart**

```python
mtcars_data["cyl"].value_counts()
```

```python
plt.pie(
    x=mtcars_data['cyl'].value_counts(),
    labels=["8 CYL","4 CYL","6 CYL"],
    explode=[0.01, 0.01, 0.01],   # use list, not set
    autopct='%1.0f%%'
)
plt.figure(figsize=(10,15))
plt.show()
```

## 2. Bivariate Analysis

**Important charts:**
* Barplot
* Scatterplot

#### This bar chart is wrong because in x-axis the numbers are spreaded so we choose the seaborn charts

```python
plt.bar(data=mtcars_data,x="am", height="mpg")
```

### 2.1 Barplot- 
* X-axis should be in **Discrete** and y-axis should be **Continous**
* When you want to compare data distribution between one **Discrete** variable VS one **Continous** variable you'l be going for **BARPLOT**

```python
sns.barplot(data=mtcars_data,x="am", y="mpg")
```

```python
mtcars_data.head(7)
```

### 2.2 Scatterplot 
If I want to compare the relationship between Two **Continous** data I'll be going for Scatterplot

#### In Matplotlib

```python
plt.scatter(x="wt", y="mpg" ,data=mtcars_data)
```

#### In Seaborn

```python
sns.scatterplot(x="wt", y="mpg" ,data=mtcars_data)
```

### Heatmap

#### Pearson Correlation ,Co-Efficient
-1 to +1

```python
del mtcars_data["model"]
```

```python
corr_matrix=mtcars_data.corr()
corr_matrix
```

```python
sns.heatmap(corr_matrix,annot=True)
plt.figure(figsize=(10,10))
plt.show()
```

#### Notes :

* 0    to 0.20 = Very Weak Correlation
* 0.21 to 0.40 = Weak Correlation
* 0.41 to 0.60 = Normal / Standard Correlation
* 0.61 to 0.80 = Strong Correlation
* 0.81 to 1    = Very strong Correlation


# Advanced Data Visualization

```python
import plotly.express as px
df = px.data.gapminder().query("year == 2007").query("continent == 'Europe'")
df.loc[df['pop'] < 2.e6, 'country'] = 'Other countries' # Represent only large countries
fig = px.pie(df, values='pop', names='country', title='Population of European continent')
fig.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load example dataset
penguins = sns.load_dataset("penguins")

# Plot histogram with KDE
sns.histplot(data=penguins, x="flipper_length_mm", kde=True, bins=20, color="skyblue")
plt.title("Flipper Length Distribution with KDE")
plt.show() 
```
