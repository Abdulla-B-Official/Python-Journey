## 1 . LIST
* list accepts Duplicate values
* list is one dimentional
* list is heterogenous data structure(ex:accepts int,string,float,etc..,)
* Indexing and slicing is possible
* Accepts Mutable (Mutable-Values once assigned to a variable can be changed inside the list)

#### 1.1 APPEND() - adds the element at the end of the list

```python
Furnitures=["Chair","Table","Window","Door"]
```

```python
type(Furnitures)
```

```python
Furnitures.append("Sofa")
Furnitures.append("2")
Furnitures
```

#### 1.2 CLEAR() - Removes all the elements from the list

```python
Furnitures.clear()
Furnitures
```

#### 1.3 COPY() - Returns a copy of the list

```python
Furnitures=["Chair","Table","Window","Door"]
Furni=Furnitures.copy()
Furni
```

#### 1.4 EXTEND() - Add the elements of a list (or any iterable), to the end of the current list

```python
Furnitures.extend("Chair")
Furnitures
```

#### 1.5 INDEX() - Returns the index of the first element with the specified value

```python
Furnitures=["Chair","Table","Window","Door"]
Furnitures.index("Window")
```

#### 1.6 INSERT() - Adds an element at the specified position

```python
Furnitures.insert(4,"Bed")
Furnitures
```

#### 1.7 POP() - Removes the element at the specified position

```python
Furnitures.pop(-1)
Furnitures
```

#### 1.8 REMOVE() - Removes the first item with the specified value

```python
Furnitures.remove("Table")
Furnitures
```

#### 1.9 REVERSE() - Reverses the order of the list

```python
Furnitures.reverse()
Furnitures
```

#### 1.10 SORT() - Sorts the list

```python
Furnitures.sort()
Furnitures
```

#### 1.11 count() - Returns the number of elements with the specified value

```python
Furnitures=["Chair","Table","Window","Door","Chair"]
Furnitures.count("Chair")
```

## 3 . SETS
* set is 1 dimenstional
* set is Heterogenous
* set don't allow duplicate values
* Indexing and slicing is not possible
* set is Mutable

```python
Furnitures={"Chair","Table","Window","Door"}
Furnitures
```

#### 3.1 ADD() - Adds an element to the set

```python
Furnitures.add("Wood")
Furnitures
```

#### 3.2 CLEAR() - Removes all the elements from the set

```python
Furnitures.clear()
Furnitures
```

#### 3.3 COPY() - Returns a copy of the set

```python
Furnitures={"Chair","Table","Window","Door","Bed"}
f=Furnitures.copy()
f
```

```python
Furnitures={"Chair","Table","Window","Door"}
```

#### 3.4 DIFFERENCE() - Returns a set containing the difference between two or more sets

```python
f.difference(Furnitures)
```

#### 3.5 DIFFERENCE_UPDATE() - Removes the items in this set that are also included in another, specified set

```python
x={"apple","mango","cherry","banana"}
y={"google","banana","microsoft","apple"}

x.difference_update(y)
print(x)
```

#### 3.6 DISCARD() - Remove the specified item

```python
Furnitures.discard("Bed")
Furnitures
```

#### 3.7 INTERSECTION() - Returns a set, that is the intersection of two other sets

```python
x={"apple","mango","cherry","banana"}
y={"google","banana","microsoft","apple"}
x.intersection(y)
```

#### 3.8 INTERSECTION_UPDATE() - Removes the items in this set that are not present in other, specified set(s)

```python
x={"apple","mango","cherry","banana"}
y={"google","banana","microsoft","apple"}
x.intersection_update(y)
print(x)
```

#### 3.9 ISDISJOINT() - Returns whether two sets have a intersection or not

```python
x={"apple","mango","cherry","banana"}
y={"google","banana","microsoft","apple"}
x.isdisjoint(y)
```

```python
x={"apple","mango","cherry","banana"}
y={"google","microsoft","intel"}
x.isdisjoint(y)
```

#### 3.10 ISSUBSET() - Returns True if all items of this set is present in another set

```python
x={"apple","mango","cherry","banana"}
y={"google","microsoft","intel"}
x.issubset(y)
```

```python
x={"apple","mango","cherry","banana"}
y={"apple","mango","cherry","banana"}
x.issubset(y)
```

```python
x={"apple","mango","cherry"}
y={"apple","mango","cherry","banana"}
x.issubset(y)
```

#### 3.11 ISSUPERSET() - Returns True if all items of another set is present in this set

```python
x={"apple","mango","cherry",}
y={"apple","mango","cherry"}
x.issuperset(y)
```

```python
x={"apple","mango","cherry","banana"}
y={"apple","mango","cherry"}
x.issuperset(y)
```

```python
x={"apple","mango","cherry","banana"}
y={"google","microsoft","intel"}
x.issuperset(y)
```

#### 3.12 POP() - Removes an element from the set

```python
x={"apple","mango","cherry","banana"}
x.pop()
x
```

#### 3.13 REMOVE() - Removes the specified element

```python
x={"apple","mango","cherry","banana"}
x.remove("apple")
x
```

#### 3.14 SYSMMETRIC_DIFFERENCE() - Returns a set with the symmetric differences of two sets

```python
x={"apple","mango","cherry","banana"}
x.symmetric_difference("apple")
```

#### 3.15 SYSMMETRIC_DIFFERENCE_UPDATE() - Inserts the symmetric differences from this set and another

```python
x={"apple","mango","cherry","banana"}
x.symmetric_difference_update("apple")
x
```

#### 3.16 UNION() - Return a set containing the union of sets

```python
x={"apple","mango","cherry","banana"}
y={"google","banana","microsoft","apple"}
x.union(y)
```

#### 3.17 UPDATE() - Update the set with the union of this set and others

```python
x={"apple","mango","cherry","banana"}
y={"google","banana","microsoft","apple"}
x.update(y)
x
```

## 4 . DICTIONARY
it is key-value Pair
* two Dimentiona
* heterogenous
* Mutable
* Indexing & Slicing is not possible
* Don't create duplicate key,but you can create duplicate values

#### 4.1 CLEAR() - Removes all the elements from the dictionary

```python
user_details={"Name":"abdulla","Age":23}
user_details.clear()
user_details
```

#### 4.2 COPY() - Returns a copy of the dictionary

```python
user_details={"Name":"abdulla","Age":23}
user_details.copy()
user_details
```

#### 4.3 FROMKEYS() - Returns a dictionary with the specified keys and value

```python
user_details={"Name":"abdulla","Age":23}
roll_no=7
user_details.fromkeys(user_details,roll_no)
```

#### 4.4 GET() - Returns the value of the specified key

```python
user_details={"Name":"abdulla","Age":23}
x=user_details.get("Name")
print(x)
```

#### 4.5 ITEMS() - Returns a list containing a tuple for each key value pair

```python
user_details={"Name":"abdulla","Age":23}
x=user_details.items()
print(x)
```

#### 4.6 KEYS() - Returns a list containing the dictionary's keys

```python
user_details={"Name":"abdulla","Age":23}
x=user_details.keys()
x
```

#### 4.7 POP() - Removes the element with the specified key

```python
user_details={"Name":"abdulla","Age":23}
x=user_details.pop("Name")
x
```

#### 4.8 POPITEM() - Removes the last inserted key-value pair

```python
user_details={"Name":"abdulla","Age":23}
x=user_details.popitem()
x
```

#### 4.9 SETDEFAULT() - Returns the value of the specified key. If the key does not exist: insert the key, with the specified value

```python
user_details={"Name":"abdulla","Age":23}
x=user_details.setdefault("Name","Fariq")
x
```

#### 4.10 UPDATE() - Updates the dictionary with the specified key-value pairs

```python
user_details={"Name":"abdulla","Age":23}
user_details.update({"Gender":"Male"})
user_details
```

#### 4.11 VALUES() - Returns a list of all the values in the dictionary

```python
user_details={"Name":"abdulla","Age":23,"Gender":"Male"}
user_details.values()
```
