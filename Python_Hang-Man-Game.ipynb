## Word Guessing Game (Mini Hangman)

### Objective:
Create a word-guessing game where the player tries to guess the letters of a hidden word
 This will test your:
1. Loops
2. Conditional logic
3. String manipulation
4. Exception handling
5. Logical thinking


### Instructions:
1.	Create a list of random words (like “python”, “machine”, “learning”, “bridge”).
2.	Randomly select one word for the game.
3.	Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
4.	Keep track of wrong guesses (max 6 attempts).
5.	End the game when the player either guesses the full word or uses all chances.


```python
import random

words=("python","machine","learning","java","laptop")
game = random.choice(words)

display=['_']*len(game)
chance=6

while chance > 0:
    for i in display:
      print(i, end=" ")
    print()
    try:
        user_input=input("Enter a letter :").lower()
        if not user_input.isalpha():
            raise ValueError("Numbers and symbols not Allowed")
        if len(user_input)!= 1:
            raise ValueError("Enter only one letter")
        if user_input in game:
            print("Correct guess!")
            for index in range(len(game)):
              if game[index]==user_input:
                display[index] = user_input
        else :
            chance -=1
            print("Wrong guess! Remaining chances:",chance)
        if '_' not in display:
           print("You win")
           break
    except ValueError as e:
        print("Invalid input:",e)
if chance==0:
    print("You Loss")
    print("The word was :",game)
```
