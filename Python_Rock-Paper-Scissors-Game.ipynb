## Rock, Paper, Scissors Game

### Objective:
Build a simple Rock-Paper-Scissors game where the player competes with the computer.
 This task will test your understanding of:
1. Input/output
2. Conditional logic
3. Random module
4. Loops
5. Exception handling


### Instructions:
1.	The user should enter their choice: rock, paper, or scissors.
2.	The computer should randomly select one of these.
3.	Compare both and print who wins.
4.	Ask if the player wants to play again- if yes, repeat; if no, exit.
5.	Handle invalid inputs gracefully using try-except or condition checks.

```python
def rock_paper_scissor_game():
    print ("WELCOME TO THE ( ROCK , PAPER , SCISSOR )GAME")
    import random
    options=("rock","paper","scissor")
    retry_options = ("yes", "no")
    Game=True
    while Game :
      try:
        user_input=input("Enter your Choice (Rock/Paper/Scissor):").lower()
          
        if user_input not in options:
         raise ValueError
          
        secret_option = random.choice(options)
            
        print("Computer's choice : ",secret_option)
            
        if user_input==secret_option:
            print(" Its a Tie ! ")
        elif(
            (secret_option == "rock" and user_input=="paper") or
            (secret_option == "paper" and user_input=="scissor") or
            (secret_option == "scissor" and user_input=="rock")
            ):
                print("You Won !! ")
        
        else :
                print("failed ! ")
        while True:      
            Try_Again = input(" Wanna Try Again ? (Yes/No) : ").lower()
            
            if Try_Again not in retry_options:
               print("Enter only (yes / no) !")
               continue
                
            if Try_Again == "no":
               print("Thank you !! ")
               Game = False
            break
      except ValueError:
        print("Enter only (Rock/Paper/Scissor)")
```

```python
rock_paper_scissor_game()
```
